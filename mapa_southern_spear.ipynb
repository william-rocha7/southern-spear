```python
# BLOCO 5: COMBATE DISTRIBUÍDO E REALISTA
import random
from datetime import timedelta, datetime

print("--- Iniciando Sequência de Combate Realista ---")

# Função auxiliar para avançar o tempo
def add_minutes(start_time_str, minutes):
    fmt = "%Y-%m-%d %H:%M:%S"
    dt = datetime.strptime(start_time_str, fmt) + timedelta(minutes=minutes)
    return dt.strftime(fmt)

# Lista de Destroyers (Quem realmente atira os Tomahawks)
# [ID, Nome, Coordenada]
destroyers = [
    ["DDG-51", "USS Arleigh Burke", COORDS["ESCORT_1"]],
    ["DDG-96", "USS Bainbridge", COORDS["ESCORT_2"]],
    ["DDG-72", "USS Mahan", COORDS["ESCORT_3"]],
    ["CG-70", "USS Lake Erie", COORDS["ESCORT_4"]]
]

# 1. ONDA DE TOMAHAWKS (Lançados pelos Destroyers)
base_time_tomahawk = "2026-01-03 02:15:00"

targets_land = [
    ["TGT_RADAR_CARLOTA", "Radar Vigilancia"],
    ["TGT_SAM_NORTE", "Bateria S-300"],
    ["TGT_FUERTE_TIUNA", "Comando Central"],
    ["TGT_PUERTO_CABELLO", "Base Naval"],
    ["TGT_MIG29_GND", "Base Aérea (Pista)"]
]

for i, tgt in enumerate(targets_land):
    # Escolhe um Destroyer aleatório para disparar
    shooter = random.choice(destroyers)
    
    # Tempo sequencial (um disparo a cada 5-10 min)
    fire_time = add_minutes(base_time_tomahawk, i * 8)
    
    # Adiciona o Míssil (Arco)
    arcs_data.append({
        "type": "MISSILE-TLAM",
        "desc": f"Tomahawk de {shooter[1]} contra {tgt[1]}",
        "start_lat": shooter[2][0], "start_lon": shooter[2][1], # Sai do Destroyer
        "end_lat": COORDS[tgt[0]][0], "end_lon": COORDS[tgt[0]][1],
        "time": fire_time
    })
    
    # Adiciona o Evento de Explosão no Alvo
    impact_time = add_minutes(fire_time, 4) # 4 min de voo
    points_data.append({
        "id": f"EXP-{i}", "label": "IMPACTO TOMAHAWK", "group": "ALVO-SEAD", "icon": "fire",
        "lat": COORDS[tgt[0]][0], "lon": COORDS[tgt[0]][1],
        "time": impact_time, "desc": f"Alvo {tgt[1]} atingido"
    })

# 2. ONDA ANTI-RADAR (HARM) - Disparados por Caças em Voo
# Precisamos criar o avião no céu antes dele atirar
base_time_air = "2026-01-03 02:40:00"

harm_targets = ["TGT_RADAR_LLANITO", "TGT_SAM_NORTE"]

for i, tgt_key in enumerate(harm_targets):
    # Posição simulada do caça (no mar, aproximando-se da costa)
    fighter_lat = 11.5 + (i * 0.2)
    fighter_lon = -66.0 + (i * 0.3)
    
    mission_time = add_minutes(base_time_air, i * 15)
    
    # A) Plota o Caça disparando (Para não sair do nada)
    points_data.append({
        "id": f"F18-SEAD-{i}", "label": f"F/A-18 Super Hornet {i+1}", 
        "group": "AEREO", "icon": "plane",
        "lat": fighter_lat, "lon": fighter_lon,
        "time": mission_time, "desc": "Disparando AGM-88 HARM"
    })
    
    # B) O Míssil sai da posição do Caça
    arcs_data.append({
        "type": "MISSILE-HARM",
        "desc": "Supressão de Radar (Magnum)",
        "start_lat": fighter_lat, "start_lon": fighter_lon, # Sai do avião
        "end_lat": COORDS[tgt_key][0], "end_lon": COORDS[tgt_key][1],
        "time": mission_time
    })

print(f"✅ Bloco 5 Realista: {len(targets_land)} Tomahawks (Destroyers) e {len(harm_targets)} HARMs (Caças) gerados.")
```
